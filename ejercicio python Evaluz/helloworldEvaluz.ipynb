```python
saludo = "Hello, World!"

## Convertir todo en mayúsculas
mayus = saludo.upper()

print(mayus)

## Convertir todo en minúsculas
minus = saludo.lower()
print(minus)

## Mostrar solo los primeros tres caracteres
tresp = saludo[:3]
print(tresp)

## Mostrar los últimos tres caracters
tresu = saludo [-3:]
print (tresu)
## Crear una lista con las vocales
def obtener_vocales(saludo):
    vocales = "aeiouAEIOU"
    return [letra for letra in saludo if letra in vocales]
resultado = obtener_vocales(saludo)
print(resultado)

## Convertir cada palabra en elemento de una lista
palabras = saludo.split()
print(palabras)
```
