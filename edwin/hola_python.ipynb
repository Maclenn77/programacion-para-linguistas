##  Funciones para cadenas (String)

```python
saludo = "Hello, World!"

## Convertir todo en mayúsculas
mayus = saludo.upper()

print(mayus)

## Convertir todo en minúsculas

## Mostrar solo los primeros tres caracteres

## Mostrar los últimos tres caracters

## Crear una lista con las vocales

## Convertir cada palabra en elemento de una lista
```

```python
saludo = "Hello, World!"
minúsculas = saludo.lower()
print(minúsculas)
```

```python
first_three_letters = saludo[0:3]
print(first_three_letters)
```

```python
last_three_letters = saludo[-3:]
print(last_three_letters)
```

```python
vocales = "aeiou"
lista_vocales = [caracter for caracter in saludo.lower() if caracter in vocales]
print(lista_vocales)
```

```python
lista_palabras = saludo.split()
print(lista_palabras)
```
