```python
from re import split
texto = "Cuando empezaba con eso, la recostaban sobre la cama o la alfombra y le colocaban un cojín bajo la cabeza. Luego era cuestión de esperar. Perla sostenía su frente mientras la abuela se zarandeaba como si quisiera escapar de su propio cuerpo. Ya casi, mamá, ya casi. A Ele le parecía curioso que su mamá llamara a la suya siempre por su nombre, Cecilia, y sólo le dijera mamá cuando no podía escucharla"

print(texto)
lista_de_palabras = texto.split()
print(lista_de_palabras)
print (len(lista_de_palabras))
preposiciones = ("con", "bajo", "de", "a", "por", "en", "para", "sobre")
lista_de_preposiciones = [caracter for caracter in texto.split() if caracter in preposiciones]
print(lista_de_preposiciones)
print(len(lista_de_preposiciones))
preposiciones_no_repetidas = set(lista_de_preposiciones)
print(preposiciones_no_repetidas)
print(len(preposiciones_no_repetidas))

diccionario = {}
diccionario["cantidad_de_palabras"] = len(lista_de_palabras)
diccionario["preposiciones_no_repetidas"] = len(preposiciones_no_repetidas)
print(diccionario)

```
