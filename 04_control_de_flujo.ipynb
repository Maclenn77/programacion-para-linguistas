## 0. Pedir datos al usuario con `input()`

En Python, la función `input()` muestra un mensaje en pantalla y espera a que la persona usuaria escriba algo y presione Enter. Lo que escribe se guarda **siempre como texto (`str`)**, incluso si son números.

```python
nombre = input("¿Cuál es tu nombre? ")
print("Hola,", nombre)
```

Si necesitas un número (como la edad), tienes que convertir el resultado con `int()`:

```python
edad = int(input("¿Cuál es tu edad? "))
```

**Pista:** para quitar espacios en blanco al inicio/final de lo que escribe la persona usuaria, usa `.strip()`. Por ejemplo, `"  ".strip()` regresa `""` (una cadena vacía).
