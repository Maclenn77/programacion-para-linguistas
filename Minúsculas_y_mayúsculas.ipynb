## Algoritmo de mayúsculas y minúsculas en Python 3

Desde Python 3, las cadenas en Python se movieron de bytes a Unicode. Esto cambia cómo se maneja la transformación a mayúsculas.

### Algoritmo simple para convertir a mayúsculas con ASCII

En ASCII*, un texto se puede transformar a mayúsculas de una manera muy sencilla: se toma su código en bytes y se le resta 32, ya que es la diferencia entre una minúscula y otra mayúscula.

```
a = 97

A = 65

97 - 32 = 65
```

Sin embargo, este truco deja de funcionar para caracteres especiales (ej. á, è, î, etc.) y caracteres fuera del alfabeto latino.


### ¿Qué cambia con Unicode?

Unicode mapea las mayúsculas (o minúsculas) asociadas a un caracter de una base de datos. Python usa este "mapa" para encontrar la mayúscula correspondiente.

### Casos especiales

Este mapeo cubre el 99% de los casos, pero aún quedan algunos casos especiales que cubrir:

- Alemán: 'ß'.upper() → 'SS' (un carácter se convierte en dos)
- Griego: la sigma final ς y σ se normaliza a Σ. De mayúsculas a minúsculas, va a variar según su posición en una palabra si Σ → ς v σ
- Turco: aquí Python usa las reglas no sensibles al locale, así que 'i'.upper() da 'I' (no 'İ', que sería la regla turca)


---
***ASCII** (American Standard Code for Information Interchange) es un estándar de codificación de caracteres que asigna números del 0 al 127 a letras, números, signos de puntuación y caracteres de control.*

````python

# Alemán
strassen = 'straßen'
print(f'En minúsculas tiene {len(strassen.lower())} caracteres y en mayúsculas tiene {len(strassen.upper())} caracteres')

# Griego
greek_mayus = "ΟΔΥΣΣΕΥΣ"
greek_minus = greek_mayus.lower()

print(f'Sigma se transforma diferente según su posición en una palabra.')
print(f'Ejemplo: En {greek_mayus}, Σ se convierte en ς v σ: {greek_minus}.')
print(f'De minúsculas a mayúsculas, tanto ς y σ se convierten en Σ.')

# Turco, no manejado por Python
# La regla indica que 'istanbul' -> 'İSTANBUL', pero Python no maneja
# este caso especial

turco = 'istanbul'
mayus_turco = turco.upper()
esperado = 'İSTANBUL'

print(f'Turco: {mayus_turco}. ¿Es lo esperado? R. {mayus_turco == esperado}')

# Para manejar el turco, se requiere una librería especial (script de abajo)
````

````python

import icu # Instalen la librería antes de correr

loc = icu.Locale('tr_TR') # Se especifica qué lengua usamos
upper = icu.UnicodeString(turco).toUpper(loc) # Especificamos que es una cadena
# unicode que queremos convertir en mayúsculas usando el locale del turco

print(f'Mayúsculas con icu = {str(upper)}. ¿Es lo esperado? R. {upper == esperado}')  # 'İSTANBUL' (con punto en la İ)
````
