## 1er programa

```python
saludo = "Hello, World!"
print(saludo)
```

## Funciones con stdtype `str`

```python
# Python 3.11.15
# cf. https://docs.python.org/3/library/stdtypes.html
```

### Convertir a mayúsculas

```python
mayús = saludo.upper()
print(mayús)
```

### Convertir a minúsculas

```python
minús = saludo.lower()
print(minús)
```

### Mostrar los 3 primeros caracteres

```python
subcadena_inicial = saludo[:3]
print(subcadena_inicial)
```

### Mostrar los 3 últimos caracteres

```python
subcadena_final = saludo[-3:]
print(subcadena_final)
```

### Convertir cada palabra en elemento de una lista

```python
palabras = saludo.split()
print(palabras)
```

#### sin puntuación usando regex

```python
import re
patrón_w = r'\w+'
re.findall(patrón_w,saludo)
```

### Filtrar por vocales y convertir en una lista

```python
patrón_v = r'[aeiou]+'
re.findall(patrón_v,saludo)
```
