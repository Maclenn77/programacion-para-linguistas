```python
parrafo = """El trabajo de Grice marca un antes y un después en nuestra comprensión del significado
en las lenguas naturales. Aunque él solo esbozó las líneas de un programa de investigación, el
desarrollo posterior de sus ideas dio origen a sofisticados trabajos y teorías sobre la manera
en la que los seres humanos logramos transmitir significados complejos a partir de los significados
lingüísticos. Uno de los más importantes es la teoría de la relevancia, de la que se ocupa este capítulo.
Antes de entrar en sus detalles, es importante recapitular brevemente el impacto de Grice en la filosofía
del lenguaje y en el surgimiento de la pragmática como una disciplina por derecho propio, con un objeto
de estudio bien definido, un marco analítico y una metodología. Contrario al clima de la época, Grice
mostró que sí es posible reconciliar la semántica de los lenguajes formales y la semántica de las 
lenguas naturales, si se distingue rigurosamente entre el significado de una oración y el significado
del hablante. La importancia del programa de investigación que Grice delineó puede apreciarse en dos
de sus efectos más importantes. Por una parte, su trabajo permitió estrechar notablemente el abismo 
entre los filósofos del lenguaje ideal, en la tradición de Frege, Russell, Carnap y Tarski, que
estudiaban el lenguaje como un sistema formal, y los filósofos del lenguaje ordinario, como el
Wittgenstein tardío, Austin y Strawson, que abordaban el uso lingüístico enfocándose en la complejidad
y sutileza de los significados transmitidos en la conversación. Grice hizo ver cómo es posible explicar
de manera sistemática el hecho de que significados lingüísticos relativamente esquemáticos y simples 
comuniquen contenidos sofisticados y diversos cuando se usan en contexto. Con ello conjuró el fantasma 
de la posible necesidad de establecer una semántica diferente para los lenguajes formales y para las
lenguas naturales.También Montague y Davidson defendieron la idea de que la semántica de las lenguas
naturales puede ser estudiada como un sistema formal. Pero fue el enfoque de Grice lo que permitió separar
de manera más nítida y fundamentada los elementos del significado comunicado que deben ser considerados
como parte de la semántica de aquellos que deben ser analizados como producto de la racionalidad de la 
conversación y de convenciones no lingüísticas.El significado de una oración y el significado del hablante 
son contenidos habitualmente diferentes. Grice entendió que el primero es solamente un vehículo para transmitir
el segundo y que este último es una intención expresada de manera abierta, que se realiza al ser reconocida.
Por eso, los enunciados no son señales que capturen de manera completa los significados que un hablante 
quiere transmitir, sino solamente evidencia que un hablante ofrece a sus oyentes para que ellos puedan 
identificarlo.Para Grice, el significado del hablante es reconocido por medio de un proceso que incluye
la descodificación del significado de la oración, pero que descansa en igual medida también en la inferencia
conversacional. En ese proceso de búsqueda del contenido que mejor explica la conducta verbal de un hablante,
los oyentes se guían por expectativas sobre ciertos estándares que se asume que los enunciados satisfagan.
Según Grice, dado que la conversación es una actividad racional, esperamos que sea cooperativa. Esta expectativa
fija un parámetro con el cual evaluar hipótesis alternativas sobre cuál es el significado que un hablante
quiere transmitir."""

#convertir parrafo a minusculas
parrafo_lower = parrafo.lower();

#separar palabra y hacer una lista con ellas
palabras = parrafo_lower.split()
#numero de palabras es igual al numero de elementos de la lista
#ojo que no separa los signos de puntuacion solo divide por espacios
num_palabras = len(palabras);

print("El parrafo tiene un total de ")
print(num_palabras)
print("palabras")

# Crear una tupla de preposiciones
tupla_preposiciones = ("a", "ante", "bajo", "cabe", "con", "contra", "de", "desde", "durante", 
"en", "entre", "hacia", "hasta", "mediante", "para", "por", "según",
"sin", "sobre", "tras")

#declarar lista de preposiciones que existen en el parrafo
lista_prep = []

#recorres la lista de palabras y verificar si hay una preposicion
#por cada elemento
for elemento in palabras:
    if elemento in tupla_preposiciones:
       lista_prep.append(elemento)

print("preposiciones")
print(lista_prep)
print("hay un total de")
print(len(lista_prep))
print("preposiciones incluyendo repetidas")

#declarar un conjunto para no incluir repeticiones
conjunto_prep = set()

#recorrer las lista de preposiciones y agregarlas al conjunto
for elemento in lista_prep:
    conjunto_prep.add(elemento)

print("preposiciones sin repetir")
print(conjunto_prep)
print("hay un total de")
print(len(conjunto_prep))
print("preposiciones SIN repetidas")

#insertar los resultados en un diccionario
#la llave seria la preposicion y el valor sera el numero de repeticiones
#en el parrafo

diccionario_prep = {}

for preposicion in lista_prep:
    if preposicion in conjunto_prep:
        # si la palabra ya existe en el diccionario, le sumamos 1 a su valor
        if preposicion in diccionario_prep:
            diccionario_prep[preposicion] = diccionario_prep[preposicion] + 1
        else:
            # si es la primera vez que aparece, la agregamos con el valor 1
            diccionario_prep[preposicion] = 1
print("frecuencia de preposiciones")
print(diccionario_prep)
```
